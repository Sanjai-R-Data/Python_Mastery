```python
# numpy operations using linear algebra
import numpy as np

a = np.array([2, 4, 6])
b = np.array([1, 3, 5])

print("Vector A:", a)
print("Vector B:", b)

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Scalar multiplication:", 2 * a)
print("Element-wise multiplication:", a * b)
print("Dot product:", np.dot(a, b))
print("Magnitude of A:", np.linalg.norm(a))
```

```python
import numpy as np

A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

B = np.array([
    [7, 8],
    [9, 10],
    [11, 12]
])

print("A shape:", A.shape)
print("B shape:", B.shape)

C = A @ B

print("Result:")
print(C)
```
