```python
# Problem 1 — Sum of List Elements
a = int(input("number of elements: "))
mylist = []
for i in range(a):
  b = int(input("enter number: "))
  mylist.append(b)
sum = 0
for i in mylist:
  sum = sum+i
print(sum)
```

```python
# Problem 3 — Find the Largest Number
a = int(input("enter number of elements: "))
mylist = []
for i in range(a):
  b = int(input("enter the elements: "))
  mylist.append(b)
mylist.sort()
print(mylist[-1])
```
